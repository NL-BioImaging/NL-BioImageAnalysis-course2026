# napari in Jupyter

This notebook shows a few examples how to run napari from a jupyter notebook.

You can find more examples here:
https://thejacksonlaboratory.github.io/intro-napari-workshop/notebooks/index.html

```python
import napari
import matplotlib.pyplot as plt
from napari.utils import nbscreenshot
from skimage.io import imread
import numpy as np
```

```python
image = imread('data/000_img.png')
viewer = napari.Viewer()
viewer.add_image(image, name='cells')
```

```python
#read back
labels = viewer.layers['committed_objects'].data
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
axes[0].imshow(image, cmap='gray',vmin=np.percentile(image, 1), vmax=np.percentile(image, 99.8))
axes[0].set_title('image')

axes[1].imshow(labels, cmap='tab20', interpolation='nearest')
axes[1].set_title(f'labels: {labels.max()} cells')
 
for ax in axes:
    ax.axis('off')
plt.show()
```

```python
labels.max()
```

```python
#remove a label
labels[labels==5] = 0
```

```python
#update layer in napari
viewer.layers['committed_objects'].data = labels
```
